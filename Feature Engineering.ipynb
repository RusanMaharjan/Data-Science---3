```python
def changeDateTime(dataframe, *args):
    for col in args:
        dataframe[col] = pd.to_datetime(dataframe[col], format='mixed')
```

```python
changeDateTime(df, 'DateofTermination', 'DOB', 'DateofHire', 'LastPerformanceReview_Date')
```

```python
df['HireYear'] = df['DateofHire'].dt.year
```

```python
df['HireMonth'] = df['DateofHire'].dt.month
df['HireDay'] = df['DateofHire'].dt.day
```

```python
## Extract date name
df['HireMonth_Name'] = df['DateofHire'].dt.month_name()
df['HireDay_Name'] = df['DateofHire'].dt.day_name()
```

### Convert Continuous data to Categorical data

```python
def convertSalary(salary):
    if salary > 0 and salary <= 70000:
        return 'Low Salary'
    elif salary > 70000 and salary <= 150000:
        return 'Medium Salary'
    else:
        return 'High Salary'
```

```python
# apply
df['Salary_Label'] = df['Salary'].apply(convertSalary)
```
