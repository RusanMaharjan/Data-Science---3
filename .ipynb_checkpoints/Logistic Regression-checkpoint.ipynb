# Logistic Regression

```python
'''
    Logistic Regression
    -----------------------
    - Classification Algorithm
        - Binary Class Classification -> Logistic Regression
        - Multi Class Classification
    
    Stratify
    -----------
    - We only need to stratify Target Data (Y).
    - To send equal % of data to both training and testing data not to make model biasness.
    - 70% -> Healthy, 30% -> Unhealthy
    
    80/20 split
    
    Training Target Data
    ----------------------
    Healthy -> 70% -> 40%
    Unhealthy -> 30% -> 40%
    
    Testing Target Data
    ---------------------
    Healthy -> 70% -> 10%
    Unhealthy -> 30% -> 10%
    
    X_train, X_test, Y_train, Y_test = train_test_split(
        X, Y, test_size=0.2, random_state=42, stratify=Y
    )
    
    Logistic Model
    --------------------
    Arguments for model
        - solver
            1. liblinear -> If our dataset is small sized (0 - 30000), 
    
            2. lbfgs (default) -> If our dataset is medium sized (30k >, below 1 lakhs), 
                - Limited-Memory-Broyden–Fletcher–Goldfarb–Shanno
    
            3. saga -> If our dataset is large sized (> 1 lakhs)
    
        - class_weight
            - None (default)
            - balanced
    
        - random_state -> 42
    
    Metrics
    ----------
    - classification report
        - Precision
        - Recall
        - F1-Score
        - Support
        - Accuracy - 80%
        - Macro Avg
        - Weighted Avg
    
    - confusion matrix
        - True Positive
        - True Negative
        - False Positive
        - False Negative
'''
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Import Logistic Regression
from sklearn.linear_model import LogisticRegression
# Import Feature Scaling Packagee
from sklearn.preprocessing import StandardScaler
# Import train test split package
from sklearn.model_selection import train_test_split
# Import classification metrics
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, roc_auc_score
```

```python
df = pd.read_csv('Dataset/Cardiovascular_Disease.csv')
df
```

## Data Cleaning

```python
df['age'] = df['age']//365
```

```python
# df['age'].max()
```

```python
# sns.boxplot(df[['age', 'height', 'weight']])
# sns.boxplot(df[['ap_hi', 'ap_lo']])
```

```python
df = df[
    (df['height'].between(145, 185)) & 
    (df['weight'].between(45, 120)) &
    (df['ap_hi'].between(90, 200)) &
    (df['ap_lo'].between(50, 90))
]
```

```python
# df
```

## Feature and Target

```python
features = ['age', 'gender', 'height', 'weight', 'ap_hi', 'ap_lo', 'cholesterol', 'gluc', 'smoke', 'alco', 'active']
target = 'cardio'
```

```python
X = df[features]
Y = df[target]
```

## Train Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42, stratify=Y
)
```

## Feature Scaling

```python
scaler = StandardScaler()
X_train_scale = scaler.fit_transform(X_train)
X_test_scale = scaler.transform(X_test)
```

## Define Model

```python
model = LogisticRegression(
    solver='lbfgs',
    class_weight='balanced',
    random_state=42
)
model.fit(X_train_scale, Y_train)
Y_pred = model.predict(X_test_scale)
```

## Metrics

```python
cr = classification_report(Y_test, Y_pred)
cm = confusion_matrix(Y_test, Y_pred)
```

```python
print('Classification Report')
print('-'*35)
print(cr)
```

```python
'''
Macro Avg
    - Metric1 + Metric2 + ... + Metric_n / n

Weighted Avg
    - metric1 * support1 + metric2 * support2 + ... +  metric_n * support_n / total_support
'''
# macro_avg = (0.76 + 0.67) / 2
# macro_avg

# weighted_avg = ((0.76 * 6631) + (0.67 * 5826))/12457
# weighted_avg
```

```python
print('Confusion Matrix')
print('-'*35)
print(cm)
```

```python
'''
- True Positive -> 5071
- False Positive -> 1560
- False Negative -> 1947
- True Negative -> 3879
'''
```

# ROC AUC Curve

```python
'''
Threshold
    - Default (0.5)

    Email spam detection

    Threshold -> 0.5
    
    TPR -> True Positive Rate
        - TP / (TP - FN)
    
    FPR -> False Positive Rate
        - FP / (FP - TN)
    
    JScore -> Best Threshold Position Find. (TPR - FPR)

'''
```

```python
# calcuation of probability
Y_prob = model.predict_proba(X_test_scale)[:,1]
Y_prob
```

```python
## Convert confusion matrix data to one dimension using ravel
tn, fp, fn, tp = cm.ravel()

# calcuate True Positive Rate and False Positive Rate
tpr_default = tp / (tp - fn)
fpr_default = fp / (fp - tn)

print(f'True Positive Rate: {tpr_default}')
print(f'False Positive Rate: {fpr_default}')
```

```python
## Find ROC Curve and ROC AUC Score data
fpr, tpr, threshold = roc_curve(Y_test, Y_prob)
auc_score = roc_auc_score(Y_test, Y_prob)

print(f'ROC False Positive Rate: {fpr}')
print(f'ROC True Positive Rate: {tpr}')
print(f'ROC Threshold: {threshold}')
```

```python
## Calcuate JScore for best threshold positions
j_score = tpr - fpr

best_idx = np.argmax(j_score)

best_tpr = tpr[best_idx]
best_fpr = fpr[best_idx]
best_threshold = threshold[best_idx]

print(f'Best Index: {best_idx}')
print(f'Best True Positive Rate: {best_tpr}')
print(f'Best False Positive Rate: {best_fpr}')
print(f'Best Threshold: {best_threshold}')
```

### ROC AUC Curve Visualization

```python
plt.plot(fpr, tpr, color='gray')
plt.plot([0, 1], [0, 1], 'k--')
plt.scatter(best_fpr, best_tpr, color='red', edgecolor='black')
```

```python
# calculation of probability of new threshold
Y_prob_opt = (Y_prob >= best_threshold).astype(int)
Y_prob_opt
```

```python
new_cr = classification_report(Y_test, Y_prob_opt)
new_cm = confusion_matrix(Y_test, Y_prob_opt)
```
