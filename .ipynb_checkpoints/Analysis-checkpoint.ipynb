```python
'''
Group by
Aggregate Functions -> (mean, max, min, count, sum)
Order by -> sort_values(ascending=True/False, by=column_name, ignore_index=True/False (default))
rename -> dictionary {'current_column_name':'replaced_column_name'}
agg -> for analysis of multiple aggregate function columns
'''
```

```python
import pandas as pd
```

```python
df = pd.read_csv('hr_cleaned_data.csv')
```

```python
df.head()
```

```python
df.columns
```

```python
# Find total employees hired through each Recruitment Source.
'''
    Syntax
    -------
    -> dataframe.groupby(column_name)[column_name].agg_function()
'''
# df['RecruitmentSource'].unique()
```

```python
emp_recruitement = df.groupby('RecruitmentSource')['EmpID'].count()
```

```python
emp_recruitement
```

```python
emp_recruitement = df.groupby('RecruitmentSource')['EmpID'].count().reset_index().rename(
    columns = {'EmpID': 'Total Employees', 'RecruitmentSource': 'Recruitment Source'}
).sort_values(ascending=False, by='Total Employees', ignore_index=True)
emp_recruitement.tail()
```

```python
emp_recruitement = df.groupby('RecruitmentSource')['EmpID'].count().rename('Total Employees').rename_axis('Recruitment Source')
format_data = emp_recruitement.to_markdown(numalign='center', stralign='center')
```

```python
print(format_data)
```

```python
# Find total employees hired through each Recruitment Source and Gender.
recruitment_gender = df.groupby(['RecruitmentSource', 'Gender'])['EmpID'].count().reset_index().rename(
    columns = {'RecruitmentSource': 'Recruitment Source', 'EmpID':'Total Employees'}
)
recruitment_gender
```
