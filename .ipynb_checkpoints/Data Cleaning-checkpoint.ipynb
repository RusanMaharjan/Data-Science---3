# Data Cleaning

```python
import pandas as pd
```

```python
df = pd.read_csv('Dataset/Employees.csv')
df.head()
```

```python
'''
Null Values
----------
NaN -> Not a Number -> Object, Int, Float, Boolean
NaT -> Not a Time -> Date / Time
'''
```

```python
'''
Process of Data Cleaning
---------------------------
1. Handle Null Value
    - Fill Null Values
        -> fillna()

        - Continuous Data -> Mean / Median
        -> Skewness -> .skew()
            -> 0 to 0.5 (Normal Distribution / Bell Curve) -> Mean
            -> Less than 0 (Left Skewness / Negative Skewness) or more than 0.5 (Right Skewness / Positive Skewness) -> Median

            -> Calculate both mean and median of data
            -> If both values are equal or almost equal -> Mean value

        -> Fill custom record if any values are not correct to fill.
        
        - Categorical Data / Discrete Data -> Mode
        
    - Drop Null Values
        -> dropna()
        -> If data consists of more than 70% null values drop column.

2. Remove Duplicate Records
    - Find duplicate records -> duplicated()
    - Remove duplicate records -> drop_duplicates()

3. Remove Unwanted Data
    a. Remove unwanted characters
        -> str.replace()
    b. Remove Whitespace
        -> str.strip()
    c. Text Transformation
        -> str.upper() / str.lower() / str.capitalize()
'''
```

## Fill Data

```python
commission_skew = df['COMMISSION_PCT'].skew()
commission_skew
```

```python
commission_mean = round(df['COMMISSION_PCT'].mean(), 2)
commission_median = df['COMMISSION_PCT'].median()

print(f'Commission Mean: {commission_mean}')
print(f'Commission Median: {commission_median}')
```

```python
# df['COMMISSION_PCT'].fillna(commission_mean, inplace=True)
# df['COMMISSION_PCT'] = df['COMMISSION_PCT'].fillna(commission_mean)
df.fillna({'COMMISSION_PCT': commission_mean}, inplace=True)
```

```python
# df['COMMISSION_PCT'].unique()
```

```python
## Fill Manager Record
```

```python
mgr_mode = df['MANAGER_ID'].mode()[0]
dep_mode = df['DEPARTMENT_ID'].mode()[0]
```

```python
print(mgr_mode)
print(dep_mode)
```

```python
df.fillna({'MANAGER_ID': mgr_mode}, inplace=True)
df.fillna({'DEPARTMENT_ID': dep_mode}, inplace=True)
```

#### Duplicate Data

```python
## Find duplicate values
# duplicated(subset = [column_name])
# drop_duplicated(subset = [column_name], keep='first'/'last')
duplicate_data = fraud[fraud.duplicated(subset=['Transaction_ID'])]
# duplicate_data
```

```python
fraud = fraud.drop_duplicates(subset=['Transaction_ID'])
```

#### Clean Data

```python
fraud['Transaction_Amount'].skew()
```

```python
median_amount = fraud['Transaction_Amount'].median()
fraud.fillna({'Transaction_Amount': median_amount}, inplace=True)
```

```python
time_mode = fraud['Time_of_Transaction'].mode()[0]
fraud.fillna({'Time_of_Transaction': time_mode}, inplace=True)
```

```python
fraud['Device_Used'].unique()
```

```python
fraud.fillna({'Device_Used': 'Unknown Device'}, inplace=True)
```

```python
fraud['Location'].unique()
```

```python
fraud.fillna({'Location': 'Unknown Location'}, inplace=True)
```

```python
fraud['Payment_Method'].unique()
```

```python
fraud.fillna({'Payment_Method': 'Unknown Method'}, inplace=True)
```

```python
# function to clean data
def clean_data(dataframe, column, data):
    dataframe.fillna({column:data}, inplace=True)
```

```python
median_amount = fraud['Transaction_Amount'].median()
time_mode = fraud['Time_of_Transaction'].mode()[0]
# fraud.fillna({'Transaction_Amount': median_amount}, inplace=True)
```

```python
clean_data(fraud, 'Transaction_Amount', median_amount)
```

```python
clean_data(fraud, 'Time_of_Transaction', time_mode)
```

## Remove Unwanted Data

```python
hr = pd.read_csv('Dataset/HR_Dataset Refresh.csv')
hr
```

```python
hr.info()
```

```python
hr.isnull().sum()
```

#### Clean Data

```python
mgr_mode = hr['ManagerID'].mode()[0]
hr.fillna({'ManagerID': mgr_mode}, inplace=True)
```

#### Duplicate Data

```python
duplicate_data = hr[hr.duplicated(subset=['EmpID'])]
# duplicate_data.head()
```

```python
# check specific duplicate values
# emp_id = int(input('Enter employee id: '))
# data = hr[hr['EmpID'] == emp_id]
# data
```

```python
hr = hr.drop_duplicates(subset=['EmpID'])
# hr
```

```python
# select_dtypes() -> select specific data types data.
# object_data = hr.select_dtypes(['object', 'int64'])
object_data = hr.select_dtypes(['object'])
object_data.columns
```

```python
# remove unwanted column
object_data = object_data.drop(columns=['DateofTermination', 'DOB', 'DateofHire', 'LastPerformanceReview_Date'])
```

```python
object_data.columns
```

```python
# object_data['Department'].unique()
```

```python
## Function to find unique values
columns = ['Employee_Name', 'Position', 'State', 'Gender', 'MaritalDesc',
       'CitizenDesc', 'HispanicLatino', 'RaceDesc', 'TermReason',
       'EmploymentStatus', 'Department', 'ManagerName', 'RecruitmentSource',
       'PerformanceScore']

def unique_values(dataframe, data_columns):
    for col in data_columns:
        print(col)
        print('-'*35)
        unique_data = dataframe[col].unique()
        print(unique_data)
        print('\n')
```

```python
# unique_values(object_data, columns)
```

#### Strip Data

```python
object_data['Department'] = object_data['Department'].str.strip()
```

```python
def removeWhiteSpace(dataframe, data_columns):
    for col in data_columns:
        dataframe[col] = dataframe[col].str.strip()
```

```python
removeWhiteSpace(hr, columns)
```

#### Replace Data

```python
'''
-> .str.replace()
Special Characters / regular expressions -> .str.replace(find, replace_value, regex=False/True (default))
Normal Text -> str.replace(find, replace_value)
'''
```

```python
hr['Employee_Name'] = hr['Employee_Name'].str.replace(',', ' ', regex=False)
```

#### Change DataType

```python
# .astype(datatype) -> int, float, boolean, object/str
hr['ManagerID'] = hr['ManagerID'].astype(int)
```

```python
## DateTime
'''
Format
--------
1. '%d.%m.%y'
2. '%d-%m-%y'
3. mixed

dayfirst=True/False (default)
'''
```

```python
# ['DateofTermination', 'DOB', 'DateofHire', 'LastPerformanceReview_Date']
# dataframe[col] = pd.to_datetime(dataframe[col], format='mixed'/dayfirst=True)
# After Conversion -> YYYY-MM-DD
hr['DOB'] = pd.to_datetime(hr['DOB'], format='mixed')
```

```python
def changeDateTime(dataframe, *args):
    for col in args:
        dataframe[col] = pd.to_datetime(dataframe[col], format='mixed')
```

```python
changeDateTime(hr, 'DateofTermination', 'DOB', 'DateofHire', 'LastPerformanceReview_Date')
```
