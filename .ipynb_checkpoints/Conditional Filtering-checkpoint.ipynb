# Conditional Filtering

```python
import pandas as pd
```

```python
df = pd.read_csv('Dataset/Employees.csv')
df.head()
```

```python
'''
Types of Data
---------------
1. Numeric (AND)
    - Discrete -> Whole Numbers (Positive / Negative) -> ID
    - Continuous -> Whole Numbers / Decimal numbers -> Salary, Temperature, Sales, Profit, Loss, PCT
    
2. Categorical (OR)
    - Nominal -> Category that can be compared. -> High, Low, Average, Grades, ..
    - Ordinal -> Category that cannot be compared. -> Gender, Caste, Colors, ..
'''
```

#### Find details of employees whose salary is in range of 10000 to 18000.

```python
# select * from employees where salary > 10000 and salary < 18000;
emp_salary = df[(df['SALARY'] >= 10000) & (df['SALARY'] <= 18000)]
emp_salary
# df[emp_salary]
```

```python
'''
Between
--------
Range -> Inclusive
'''
```

```python
emp_sal_between = df[df['SALARY'].between(10000, 18000)]
emp_sal_between
```

#### Find details of employees who works under job AD_VP / IT_PROG.

```python
emp_job = df[(df['JOB_ID'] == 'AD_VP') | (df['JOB_ID'] == 'IT_PROG')]
emp_job
```

```python
emp_job_in = df[df['JOB_ID'].isin(['AD_VP', 'IT_PROG'])]
emp_job_in[['FIRST_NAME', 'LAST_NAME', 'JOB_ID']]
```

#### Find details of employees who works under job IT_PROG and whose salary  is in range of 4000 to 7000.

```python
emp_detail = df[(df['JOB_ID'] == 'IT_PROG') & (df['SALARY'].between(4000, 7000))]
emp_detail
```
